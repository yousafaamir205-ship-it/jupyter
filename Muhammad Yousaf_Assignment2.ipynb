```python
employee = pd.DataFrame({
    "Emp_ID":[101,102,103,104],
    "Name":["Ali","Sara","Ahmed","Fatima"]})
salary = pd.DataFrame({
    "Emp_ID":[101,102,104,105],
    "Salary":[50000, 65000,50000,45000]
})
```

```python
inner_merge = pd.merge(
    employee,salary,
    on="Emp_ID",how="inner")

print(inner_merge)
```

```python
left_merge=pd.merge(
    employee,salary,on="Emp_ID",how="left")

print(left_merge)
```

```python
left_merge=pd.merge(
    employee,salary,on="Emp_ID",how="left")
print(left_merge)
```

```python
department=pd.DataFrame({
    "Emp_ID":[101,102,103,104],
    "Department":["HR","IT","Sales","Finance"]
})
```

```python
final_m=pd.merge(
    employee,
    salary,
    on="Emp_ID",
    how="outer")

final_m= pd.merge(
    final_m,
    department,
    on="Emp_ID",
    how="outer"
)

print(final_m)
```
