```python
import random

def fitness(x):
    return x**2

def int_to_bin(x):
    return format(x, "05b")

def bin_to_int(b):
    return int(b, 2)

def init_population(size):
    return [int_to_bin(random.randint(0, 31)) for _ in range(size)]

def selection(population):
    fitness_values = [fitness(bin_to_int(ind)) for ind in population]
    total = sum(fitness_values)
    if total == 0:
        return random.choices(population, k=2)
    probs = [f / total for f in fitness_values]
    parent1 = random.choices(population, weights=probs, k=1)[0]
    parent2 = random.choices(population, weights=probs, k=1)[0]
    return parent1, parent2

def crossover(p1, p2, point=None):
    if point is None:
        point = random.randint(1, min(len(p1), len(p2)) - 1)
    return p1[:point] + p2[point:], p2[:point] + p1[point:]

def mutate(chromosome, rate=0.1):
    chromosome_list = list(chromosome)
    for i in range(len(chromosome_list)):
        if random.random() < rate:
            chromosome_list[i] = "1" if chromosome_list[i] == "0" else "0"
    return "".join(chromosome_list)

def genetic_algorithm(pop_size=6, generations=20):
    population = init_population(pop_size)
    best_fitness = []
    for gen in range(generations):
        new_population = []
        for _ in range(pop_size // 2):
            p1, p2 = selection(population)
            c1, c2 = crossover(p1, p2)
            c1, c2 = mutate(c1), mutate(c2)
            new_population.extend([c1, c2])
        population = new_population
        best = max(fitness(bin_to_int(ind)) for ind in population)
        best_fitness.append(best)
        best_individual = max(population, key=lambda x: fitness(bin_to_int(x)))
        print(f"Generation {gen + 1}: Best Fitness = {best}, Best Individual = {best_individual}")
    return best_fitness
```
