```python
import math

def alphabeta(depth, node_index, maximizing_player, values, alpha, beta, max_depth):
    if depth == max_depth:
        return values[node_index]

    if maximizing_player:
        best = -math.inf
        for i in range(2):
            val = alphabeta(depth + 1, node_index * 2 + i, False, values, alpha, beta, max_depth)
            best = max(best, val)
            alpha = max(alpha, best)
            if beta <= alpha:
                print(f"pruned at depth {depth}, node {node_index}")
                break
        return best

    best = math.inf
    for i in range(2):
        val = alphabeta(depth + 1, node_index * 2 + i, True, values, alpha, beta, max_depth)
        best = min(best, val)
        beta = min(beta, best)
        if beta <= alpha:
            print(f"pruned at depth {depth}, node {node_index}")
            break
    return best

max_depth = 3
print("The optimal value is:", alphabeta(0, 0, True, [3, 5, 6, 9, 1, 2, 0, -1], -math.inf, math.inf, max_depth))
```
