Implement the Water Jug Problem using a 4-gallon jug X and a 3-gallon jug Y.   Start with both jugs empty: (0, 0).   Use the allowed operations: fill a jug, empty a jug, and pour one jug into the other.   Find a sequence of operations that produces exactly 2 gallons in jug X.   Display the state (X, Y) after every operation.   Stop when the goal state X = 2 is reached and display the total number of steps.   

```python
X=0
Y=0
n=0
c=input("enter your choice limit for x is 4, and for y is 3")
if(c=="Fill X"):
  amount=int(input("enter your amount to fill: "))
  if(amount!=4):
    print("invalid amount")
  else:
    X=amount
  n+=1
  print(X,Y)
elif(c=="Fill Y"):
  amount=int(input("enter your amount to fill: "))
  if(amount!=3):
    print("invalid amount")
  else:
    Y=amount
  n+=1
  print(X,Y)
if(c=="Empty X"):
  X=0
  print(X,Y)
  n+=1
if(c=="Empty Y"):
  Y=0
  print(X,Y)
  n+=1
if(c=="Put X in Y"):
  Y=3
  X=1
  print(X,Y)
  n+=1
if(c=="Put Y in X"):
  Y=0
  X=3
  print(X,Y)
  n+=1

print("total steps: ",n)
```
