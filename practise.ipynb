```python
import random

guess=random.randint(1,20)

while True:
    rn=int(input("enter a random number: "))

    if guess==rn:
        print("found")
        break
    elif guess>rn:
        print("too low")
    else:
        print("too high")
        
print(guess)
```

```python
password=input("enter a password: ")
has_upper=False
has_lower=False
has_digital=False
has_special=False
special = "!@#$%^&*()_+-=[]{}|;:',.<>?/"
for ch in password:
    if ch.isupper():
        has_upper = True
    elif ch.islower():
        has_lower = True
    elif ch.isdigit():
        has_digital = True
    elif ch in special:
        has_special = True
        
if has_upper and has_digital and has_lower and has_special and len(password)>=8:
    print("strong password")
else:
    print("weak password")
```

```python
number=[1,2,2,3,4,5,6,5]

seen=[]

duplicate=[]
n=[]
for num in number:
    if num in seen and num not in duplicate:
        duplicate.append(num)
    elif num not in seen and num  in duplicate:
        n.append(num)
        
    else:
        seen.append(num)
print("duplicate: ",duplicate)

print("seen: ",seen)
print("number not duplicate",n)
```
