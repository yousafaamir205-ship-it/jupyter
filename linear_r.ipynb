implemnting linear regression without using any library

```python
X=[1,2,3,4,5]
Y=[2,4,5,4,5]
mean_x=sum(X)/len(X)
mean_y=sum(Y)/len(Y)
numerator=sum((x_i-mean_x)*(y_i-mean_y)for x_i,y_i in zip(X,Y))
denominator=sum((x_i-mean_x)**2 for x_i in X)
m=numerator/denominator
b=mean_y-(m*mean_x)
print(f"slope:{m:.4f}")
print(f"Intercept:{b:.4f}")
print(f"Equation:y={m:.4f}x+{b:.4f}")
def predict(x_val):
    return m*x_val+b

print(f"Predicted value for x=6: {predict(6):.4f}")
```

Linear regression using sklearn


```python
import numpy as np
from sklearn.linear_model import LinearRegression
X=np.array([[1.0,2],[1.5,3],[2.0,3],[2.5,4],[3.0,5]])
y=np.array([2.0,3.1,3.8,4.9,5.8])
model=LinearRegression()
model.fit(X,y)
print("weights: ",np.round(model.coef_,4))
print("Bias",round(model.intercept_,4))
new_house=np.array([[2.2,3]])
prediction=model.predict(new_house)
print(f"Predicted Price: ${prediction[0]*100000:,.2f}")
```
