```python
import matplotlib.pyplot as plt
# pyplot plot is function that is necessory for matplotlib
```

```python
x=[1,2,3,4,5]
y=[10,20,30,40,50]
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y, color="r", marker= "<", linestyle="--", linewidth=2, markersize =12)
```

```python
# giving names to chart and axis
plt.plot(x,y, color="r", marker= "<", linestyle="--", linewidth=2, markersize =12)
plt.title("my first chart via python")
plt.xlabel("x axis numbers")
plt.ylabel("y axis numbers")
```

```python
# amking charts of multiple variables and adding legend
x=[10,20,30,40]
y=[1,2,3,4]
z=["A","B","C","D"]

plt.plot(x,y, label="sales 2020")
plt.plot(x,z, label="sales 2026")
plt.title("my second chart via python")
plt.xlabel("x axis numbers")
plt.ylabel("y axis numbers")
plt.legend()
plt.show()
```

```python
x=[1,2,3,4]
y=[123,220,340,450]
plt.bar(x,y)
plt.title("my first bar chart in python")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.show()
```

```python
# histogram

import random
data=[random.randint(1,20) for _ in range(500)]
# bins means number of bars in a histogram
plt.hist(data,bins=20)
plt.title(" my first histogram in python")
plt.show()
```

```python
# pie chart

categories=["a","b","c","d","e"]
sales=[10,20,30,40,50]
# autopct is total percentage, f is float and 1.1 is 1 decimal place
plt.pie(sales,labels=categories,autopct='%1.1f%%', startangle=90)
plt.title("my first pie chart in python")
plt.show()
```

```python
# scatter plot
y1=[1,2,3,49,5]
y2=[10,20,33,40,50]
plt.scatter(y1,y2)
plt.title("my first scatter plot")
plt.show()
```

```python
# sub plot
# data1
categories=["mon","tues","wed","thurs","fri"]
sales=[10,20,30,80,99]
# data2
y1=[1,2,3,49,5]
y2=[10,20,33,40,50]

plt.figure(figsize=(12,10))
# 1st subplot
#(1,2,1) mean 1 row and 2 columns and 1 number of chart
plt.subplot(2,2,1)
plt.bar(categories,sales)
plt.title("bar chart")
plt.xlabel("x axis bar chart")
plt.ylabel("y axis of bar chart")

# 2st subplot
#(1,2,1) mean 1 row and 2 columns and 2 number of chart
plt.subplot(2,2,2)
plt.plot(y1,y2)
plt.title("line chart")
plt.xlabel("x axis line chart")
plt.ylabel("y axis of line chart")
# 3rd subplot
plt.subplot(2,2,3)
y1=[1,2,3,49,5]
y2=[10,20,33,40,50]
plt.scatter(y1,y2)
plt.title("my first scatter plot")
# 4th subplot
plt.subplot(2,2,4)
categories=["a","b","c","d","e"]
sales=[10,20,30,40,50]
# autopct is total percentage, f is float and 1.1 is 1 decimal place
plt.pie(sales,labels=categories,autopct='%1.1f%%', startangle=90)
plt.title("my first pie chart in python")


plt.show()
```
